## Part B — Fixed values and value flow

Start with one line and read it slowly:

```python
unit_price = 250.0
```

- `unit_price` is a **variable name** chosen by the programmer.
- `250.0` is a **literal value** written directly in the code.
- `=` is the **assignment operator**. Here it means “store the value on the right using the name on the left.”
- The complete line is an **assignment statement**—an instruction Python executes.

Python evaluates the **right-hand side first**, then creates or updates the name on the left. Read the line as: “Assign the value `250.0` to `unit_price`.” Do not read `=` as “is mathematically equal to”; Python uses `==` for comparison.

Meaningful `snake_case` names such as `unit_price` are easier to understand than names such as `x1`.

````python
# Source code: these are instructions written by the programmer.
unit_price = 250.0
print(unit_price)
````

The source line creates a value in the running notebook state. `print(unit_price)` then sends a readable copy of that value to the **output area**. The output `250.0` is not another instruction and editing displayed output does not change the stored value.

`print(...)` is a built-in Python tool. The parentheses hold the value we want it to display. We can use built-in tools today without learning how to create our own functions; creating functions belongs to Session 08.

### A name can be reassigned

Assignment can create a name or replace the value currently stored under that name. Predict both outputs before running.

````python
practice_quantity = 4
print(practice_quantity)

practice_quantity = 5
print(practice_quantity)
````

The second assignment does not create a second `practice_quantity`. It updates the value the runtime will use from that point onward. This is why execution order matters.

Also notice the difference between names and text:

- `customer_name` without quotes is a variable name.
- `"Maya Patel"` with quotes is a string value.

Comments begin with `#`. Python ignores the comment text; it is guidance for human readers.

## Part C — Types and operators

The scenario needs four core types:

- `int`: whole numbers such as a quantity;
- `float`: decimal numbers such as a unit price or percentage;
- `str`: text such as a customer name;
- `bool`: `True` or `False`, such as a comparison result.

Use `type()` while learning when the type is not obvious. You do not need to print it after you understand the value.

````python
print(type(quantity))
print(type(unit_price))
print(type(customer_name))
print(type(subtotal >= large_order_threshold))
````

### Assignment is not comparison

- `=` assigns a value: `quantity = 4`
- `==` compares two values and produces a boolean: `quantity == 4`

Predict each result before running.

````python
print(quantity == 4)
print(quantity != 4)
print(subtotal >= large_order_threshold)

price_is_positive = unit_price > 0
quantity_is_positive = quantity > 0
inputs_look_positive = price_is_positive and quantity_is_positive

print(inputs_look_positive)
print(not inputs_look_positive)
````

Arithmetic operators in this practical are `+`, `-`, `*`, and `/`. Parentheses make intent visible even when Python could apply precedence rules without them.

````python
ten_percent_as_decimal = discount_pct / 100
check_discount = subtotal * (discount_pct / 100)
print(ten_percent_as_decimal)
print(check_discount)
````

## Part E — Strings and indexing

Strings are quote-delimited text. For this utility, we need only a few purposeful operations:

- `.strip()` removes surrounding whitespace;
- `.title()` gives names title-style capitalisation;
- `len()` counts characters;
- indexing selects one character;
- an f-string combines labels and values readably.

````python
customer_name = customer_name.strip().title()
product_name = product_name.strip().title()

print(customer_name)
print(product_name)
print("Customer-name length:", len(customer_name))
````

Python string positions start at `0`. Therefore, `[0]` means the first character and `[1]` means the second. Negative positions count from the end: `[-1]` is the last character and `[-2]` is the second-last. Spaces are characters too.

For `Maya Patel`, the checkpoints are:

| Expression | Character |
|---|---|
| `customer_name[0]` | `M` |
| `customer_name[1]` | `a` |
| `customer_name[-2]` | `e` |
| `customer_name[-1]` | `l` |

Predict all six printed characters before running the next cell.

````python
print(customer_name[0])
print(customer_name[1])
print(customer_name[-2])
print(customer_name[-1])
print(product_name[0])
print(product_name[-1])
````
