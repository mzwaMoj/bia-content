# Part D — Tuples, sets, and intentional structure choice

Use a **tuple** for a small ordered group whose top-level positions should not be reassigned. Tuple contents are not described by the absolute claim “nothing inside can ever change,” and not every tuple is hashable.

Use a **set** when unique membership matters. A set has no positional indexing, so do not rely on its display order.

```python
order_key = ("DEMO-01", "West")
order_id, region = order_key

print("First tuple item:", order_key[0])
print("Unpacked order ID:", order_id)
print("Unpacked region:", region)
```

```python
regions_with_duplicates = ["West", "North", "West", "East", "North"]
unique_region_demo = set(regions_with_duplicates)

print("Original list:", regions_with_duplicates)
print("Unique membership:", unique_region_demo)
print("Is West present?", "West" in unique_region_demo)
```

The practical uses simple **relative filenames**. Python looks for an input relative to the notebook runtime's current working folder. Keeping the notebook and CSV together makes the learner path easy to inspect.

```python
INPUT_FILE = "S09_orders.csv"
ENRICHED_JSON_FILE = "S09_enriched_orders.json"
SUMMARY_FILE = "S09_summary.txt"
PRACTICE_TEXT_FILE = "S09_file_practice.txt"
LARGE_ORDER_THRESHOLD = 950.0

print("Expected input file:", INPUT_FILE)
print("Planned outputs:", ENRICHED_JSON_FILE, "and", SUMMARY_FILE)
```

### A small text write/read cycle

`with open(...)` closes the file when the indented block finishes, including when an error interrupts the block. Use an explicit UTF-8 encoding for predictable text handling.

```python
practice_message = "A file persists after this cell finishes."

with open(PRACTICE_TEXT_FILE, mode="w", encoding="utf-8") as text_file:
    text_file.write(practice_message)

with open(PRACTICE_TEXT_FILE, mode="r", encoding="utf-8") as text_file:
    reloaded_message = text_file.read()

print(reloaded_message)
```
