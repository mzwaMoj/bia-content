# Part B — Repetition and loop control

Before choosing loop syntax, decide **why** the repetition exists.

| Need | Natural starting choice |
|---|---|
| Repeat a known/bounded number of times | `for` |
| Repeat until a condition changes | `while` |

Both loops can use `break` and `continue`, but those keywords mean different things:

- `continue` → stop the **current iteration** and move to the next one.
- `break` → exit the **current loop** completely.

### Trace a bounded `for` loop

Before running the next cell, write the values that `attempt` will take.

````python
for attempt in range(1, 4):
    print("Attempt number:", attempt)
````

### Guided activity — bounded retry with `continue` and `break`

Run the next cell and try these behaviors:

1. Submit a blank value first — the loop should `continue` to another attempt.
2. Enter a non-blank reference — the loop should `break` immediately.
3. Rerun and submit blanks for all three attempts — the loop should end naturally after the bound is exhausted.

Notice that we are not storing attempts in a collection. The program handles one attempt at a time.

````python
accepted_reference = False

for attempt in range(1, 4):
    reference_code = input(f"Attempt {attempt}/3 - enter a reference code: ").strip()

    if reference_code == "":
        print("Reference cannot be blank. Try again.")
        continue

    print("Accepted reference:", reference_code)
    accepted_reference = True
    break

if not accepted_reference:
    print("No valid reference was entered in three attempts.")
````

### Condition-driven repetition with `while`

A customer name should not be blank. We do not know in advance how many attempts a user will need, so a `while` loop is a natural fit.

Run the cell and press Enter once without typing a name, then enter a valid name.

````python
customer_name = ""

while customer_name == "":
    customer_name = input("Customer name: ").strip()

    if customer_name == "":
        print("Customer name cannot be blank.")

print("Accepted customer:", customer_name)
````

### A safe `while` trace

A `while` loop needs a condition that can eventually become `False` or another reachable exit such as `break`.

Predict the three printed values before running.

````python
counter = 1

while counter <= 3:
    print("Counter:", counter)
    counter = counter + 1
````

### Diagnose an infinite-loop pattern

Do **not** run this defective pattern as written:

```python
counter = 1
while counter <= 3:
    print(counter)
```

Ask two questions:

1. Which value controls the loop condition?
2. Where does that value change so the condition can eventually become `False`?

Here, `counter` never changes, so the condition stays true. The fix is not exception handling; the program logic itself must be corrected.

# Part C — Functions as named responsibilities

A function is useful because it gives one responsibility a clear name and interface. Functions support:

- readability;
- reuse;
- testing;
- smaller problem-solving steps;
- passing results into other steps.

The goal is not "as many functions as possible." The goal is **clear responsibility and visible value flow**.

### Define, call, parameter, argument, return

In the function below:

- `unit_price` and `quantity` are **parameters** in the definition.
- `250.0` and `4` are **arguments** in the call.
- `return subtotal` hands the computed value back to the caller.

````python
def calculate_subtotal(unit_price, quantity):
    subtotal = unit_price * quantity
    return subtotal

subtotal = calculate_subtotal(250.0, 4)
print("Subtotal returned to caller:", subtotal)
````

### `print()` is not `return`

Predict the two values assigned to `printed_result` and `returned_result`.

A function can print something and still return nothing useful to its caller.

````python
def display_subtotal(unit_price, quantity):
    subtotal = unit_price * quantity
    print("Displayed subtotal:", subtotal)


def return_subtotal(unit_price, quantity):
    subtotal = unit_price * quantity
    return subtotal

printed_result = display_subtotal(250.0, 4)
returned_result = return_subtotal(250.0, 4)

print("Value received from display_subtotal:", printed_result)
print("Value received from return_subtotal:", returned_result)
````

### Default parameter value

A default is useful when one value is common but callers should still be able to override it.

For this session, the familiar large-order threshold is `950.0`.

````python
def classify_order(final_total, threshold=950.0):
    if final_total >= threshold:
        return "large"
    else:
        return "standard"

print("945.0 ->", classify_order(945.0))
print("1181.25 ->", classify_order(1181.25))
print("945.0 with threshold 900.0 ->", classify_order(945.0, 900.0))
````

# Part D — Function composition and scope

Real programs become easier to reason about when one function can use the result of another.

**Function composition** means that values move through several named responsibilities instead of living in one long block of code.

````python
def clean_text(text):
    return text.strip()


def build_order_label(customer_name, product_name):
    clean_customer = clean_text(customer_name).title()
    clean_product = clean_text(product_name).title()
    return clean_customer + " - " + clean_product

order_label = build_order_label("  maya patel  ", " analytics starter kit ")
print(order_label)
````

### Practical scope: local values and a module-level constant

A variable created inside a function is normally **local** to that function call.

A value defined at the notebook/script level is in module/global scope. For a fixed configuration value, an uppercase name is a common convention.

Use global/module values sparingly. Important working data should usually move through **parameters and return values** so the flow stays visible.

````python
LARGE_ORDER_THRESHOLD = 950.0


def build_order_message(final_total, threshold=LARGE_ORDER_THRESHOLD):
    # order_label is local to this function call.
    if final_total >= threshold:
        order_label = "large"
    else:
        order_label = "standard"

    return "Order classification: " + order_label

print(build_order_message(945.0))
print(build_order_message(1181.25))
````

### Scope check — can a local variable be used outside its function?

A local variable exists **inside the function call that created it**. It is not automatically available in the outer notebook/script scope.

# Part E — Expected failure and input validation

User input creates two different kinds of problems, and they should be handled differently.

| Situation | Example | Natural tool |
|---|---|---|
| Text cannot be converted to the expected number | `int("four")` | `try / except ValueError` |
| Conversion succeeds but the value violates a rule | quantity is `0` or `-2` | normal `if / elif / else` |

Exception handling is for **expected runtime failures**, not for hiding programming mistakes.

### `try`, specific `except`, `else`, and `finally`

The first example intentionally uses text that cannot become an integer.

- `except ValueError` handles the expected conversion failure.
- `else` runs only when the `try` block succeeds.
- `finally` runs whether the conversion succeeds or fails.

````python
raw_quantity = "four"

try:
    quantity = int(raw_quantity)
except ValueError:
    print("Could not convert the quantity to an integer.")
else:
    print("Converted quantity:", quantity)
finally:
    print("Conversion attempt finished.")
````

Now run the same flow with valid numeric text. Predict which blocks will run.

````python
raw_quantity = "5"

try:
    quantity = int(raw_quantity)
except ValueError:
    print("Could not convert the quantity to an integer.")
else:
    print("Converted quantity:", quantity)
finally:
    print("Conversion attempt finished.")
````

# Part F — From notebook work to script structure

A notebook is an interactive work surface. A clean `.py` script usually groups the program in a more deliberate order:

```text
configuration / constants
helper function definitions
main() definition
if __name__ == "__main__":
    main()
```

The final line is an **execution condition**: run `main()` when the file is being executed directly.

The deeper mechanics of importing code across files come later. For this session, focus on recognizing the structure and why `main()` gives the program one clear coordination point.
