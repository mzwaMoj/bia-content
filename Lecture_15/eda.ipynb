# Replace missing values

```python
df_weather['Temperature'].mean()
```

```python
df_weather['Temperature']
```

```python
# numerical variables: mean


df_weather['Temperature'] = df_weather['Temperature'].fillna(df_weather['Temperature'].mean())
```

```python
cols = df_weather.columns.tolist()
```

```python
df_weather['Event'].dtype == float
```

```python
df_weather['Event'].dtype
```

```python
cols
```

```python
df_weather['Event'].mode()[0]
```

```python
df_weather['VisibilityMiles'].mode()
```

```python
for column in cols[:]:
    if df_weather[column].dtype == float or df_weather[column].dtype == int:

        df_weather[column] = df_weather[column].fillna(df_weather[column].mean())
    else:
        df_weather[column] = df_weather[column].fillna(df_weather[column].mode()[0])

```

```python


x_axis = df_weather['Date']
y_axis = df_weather['Temperature']

plt.figure(figsize = (15, 4))
plt.plot(x_axis,y_axis, label = 'Temprature', )
plt.xlabel("date")
plt.ylabel("temp")
plt.title("Temp over time")

DewPoint = df_weather['DewPoint']

plt.plot(x_axis,DewPoint, label = 'DewPoint', linestyle ="--")
plt.xlabel("date")
plt.ylabel("temp")
plt.title("Temp over time")
plt.legend()
```
